### Forces Affecting Our Spacecraft

**Newton’s First Law** tells us that *"a body in motion continues in motion at constant velocity unless acted upon by external forces"*.

So, what are the external **forces** that act on our spacecraft? We consider just three:  
1) **Gravitational** forces (independently from both the Earth and the Moon).  
2) **Drag** forces from moving through the atmosphere.  
3) **Thrust** forces from our spacecraft’s engines.  

To keep things simple for our simulation, we do not explicitly model the mass of our spacecraft (or how it changes throughout the mission).
Instead, we consider only the **acceleration** that results from these forces.

At each point in time, based on the spacecraft’s current position and velocity, we compute independently:
- the **gravitational** acceleration (caused by both the Earth and the Moon),
- the **drag** acceleration, and
- the acceleration due to our spacecraft’s engine **thrust**.

### Gravitational Acceleration

The direction of the gravitational acceleration is from the spacecraft toward the celestial body. 
Its magnitude is inversely proportional to the square of the distance between the spacecraft and the body, and directly proportional to the product of the celestial body’s mass and the universal gravitational constant $G$. 
In astrodynamics, these are commonly combined into a single gravitational parameter $\mu$, because using a single value avoids introducing additional floating-point error from separately representing and multiplying two quantities.

<div style="padding-left: 2em; max-width: 400px;">
    $\displaystyle \mathbf{r} = \mathbf{r}_b - \mathbf{r}_s$
    <div style="display: flex; justify-content: space-between; margin-top:15px">
        $\displaystyle \mathbf{a}_g = \frac{\mu}{\left\lVert \mathbf{r} \right\rVert^2} \, \widehat{\mathbf{r}}$
        <span style="color: blue; font-weight: bold;">(2)</span>
    </div>
</div>

<div>where:</div>
<div style="padding-left:2em">
$\mathbf{a}_g$ : gravitational acceleration vector acting on the spacecraft  
    
$\mathbf{r}_s$ : spacecraft position vector  
$\mathbf{r}_b$ : celestial body position vector  
$\mu$ : standard gravitational parameter of the celestial body, defined as $\mu = G M$  
$G$ : the universal gravitational constant, $\approx 6.67 \times 10^{-11}$  
$M$ : the mass of the celestial body  
</div>

### Drag Acceleration

To compute the drag acting on the spacecraft, we must first determine its velocity relative to the surrounding atmosphere at its current position. This is called the **relative velocity**, defined as the difference between the spacecraft velocity and the local atmospheric velocity. The direction of the drag acceleration is always opposite to this relative velocity. Its magnitude depends on the atmospheric density at the spacecraft’s current altitude and increases with the square of the spacecraft’s speed relative to the atmosphere.

In aerodynamics, drag depends on several properties of the spacecraft, including its cross-sectional area $A$, mass $m$, and drag coefficient $C_d$, which reflects how streamlined the spacecraft is. These properties change during a mission, especially when stages of the spacecraft separate and are jettisoned. For the purposes of this assignment, we combine these quantities into a single drag parameter $k_d = \frac{C_d A}{m}$. This simplifies the model and avoids the need to separately track changing physical properties, while still capturing the overall influence of drag on the spacecraft’s motion.

<div style="padding-left: 2em; max-width: 400px">
    <div style="display: flex; justify-content: space-between; margin-top:15px; margin-bottom:15px">    
        $\displaystyle \mathbf{v}_{\text{rel}} = \mathbf{v}_s - \mathbf{v}_{\text{atm}}$
        <span style="color: blue; font-weight: bold;">(3)</span>
    </div>
    <div style="display: flex; justify-content: space-between; margin-top:15px; margin-bottom:15px">
        $\displaystyle \mathbf{a}_d = -\left(\frac{1}{2}\,\rho\,\frac{C_d A}{m}\,\|\mathbf{v}_{\mathrm{rel}}\|^2\right)\widehat{\mathbf{v}}_{\mathrm{rel}}$
        <span style="color: blue; font-weight: bold;">(4)</span>
    </div>
    $\displaystyle k_d = \frac{C_d A}{m}$
</div>

<div>
<div>where:</div>
<div style="padding-left:2em">
$\mathbf{a}_d$ : drag acceleration vector acting on the spacecraft  

$\mathbf{v}_s$ : spacecraft velocity vector  
$\mathbf{v}_{\text{atm}}$ : atmospheric velocity vector at the spacecraft position  
$\mathbf{v}_{\text{rel}}$ : velocity of the spacecraft relative to the atmosphere  
$\rho$ : atmospheric density at the spacecraft *altitude*  
$C_d$ : drag coefficient  
$A$ : reference (cross-sectional) area of the current spacecraft configuration  
$m$ : current mass of the spacecraft  
$k_d$ : drag parameter combining aerodynamic properties and mass
</div>

### Engine Thrust Acceleration

The thrust generated by the spacecraft’s engines produces an acceleration that is controlled both in magnitude and direction. The magnitude of the thrust acceleration is specified directly and varies over time depending on the current phase of the mission.

For most of the mission, the spacecraft’s engines operate in **prograde** mode, meaning the spacecraft is assumed to be oriented in the same direction that it is travelling. With the engines pointing directly backwards, the resulting thrust (and therefore acceleration) is in the direction of the current velocity. However, at liftoff, this assumption does not hold. The spacecraft already has an initial velocity due to the rotation of the Earth, and this velocity is entirely horizontal (eastward), while the spacecraft is initially pointing vertically upward. As a result, the initial thrust is vertical rather than aligned with the direction of motion.

In real launch missions, a pitch manoeuvre is performed shortly after liftoff to gradually reorient the spacecraft. This causes the thrust direction to transition smoothly from vertical ascent to horizontal (prograde) flight as orbital velocity builds. This behaviour is known as a *gravity turn*. 
To capture this behaviour in our simplified model, we introduce a **gravity turn mode**. In this mode, the thrust direction is computed as a blend of two directions: the vertical “up” direction (given by the position vector $\mathbf{r}_s$) and the forward direction (given by the velocity vector $\mathbf{v}_s$). The balance between these directions is controlled by a blending factor $\beta$, which increases with altitude. This causes the thrust direction to gradually rotate from vertical to prograde as the spacecraft climbs.

In **off** mode, the engine is not firing so no thrust is produced. 

The final thrust acceleration vector is obtained by multiplying the current thrust acceleration magnitude by the normalised thrust direction vector.

<div style="padding-left: 40px; max-width: 400px;">
    <div style="display: flex; justify-content: space-between; margin-bottom: 15px;">
        <span>$\displaystyle \mathbf{d}_t =
            \begin{cases}
                \mathbf{v}_s & \text{(Prograde mode)} \\[8pt]
                \mathrm{blend}(\widehat{\mathbf{r}_s}, \widehat{\mathbf{v}_s}) & \text{(Gravity turn mode)} \\[8pt]
                \mathbf{0} & \text{(Off mode)}
            \end{cases}$</span>
        <span></span>
    </div>
    <div style="display: flex; justify-content: space-between; margin-top: 15px">
        $\displaystyle \mathrm{blend}(\mathbf{u}, \mathbf{f}) = (1 - \beta) \, \mathbf{u} + \beta \, \mathbf{f}$
        <span style="color: #1a4cff; font-weight: bold;">(5)</span>
    </div>
    <div style="display: flex; justify-content: space-between; margin-top: 15px">
        $\displaystyle \beta = 1 - e^{-h / \kappa}$
    </div>
    <div style="display: flex; justify-content: space-between; margin-top: 15px">
        <span>$\displaystyle \mathbf{a}_t = a_t \, \widehat{\mathbf{d}_t}$</span>
        <span style="color: #1a4cff; font-weight: bold;">(6)</span>
    </div>
</div>

<div>
<div>where:</div>
<div style="padding-left:2em">
$\mathbf{a}_t$ : thrust acceleration vector acting on the spacecraft  

$a_t$ : current engine acceleration magnitude  
$\mathbf{d}_t$ : computed thrust direction vector  
$\mathbf{v}_s$ : spacecraft velocity vector  
$\mathbf{r}_s$ : spacecraft position vector (since position is relative to the Earth’s centre, this also defines the “up” direction)  
$\beta$ : current gravity turn blending factor  
$h$ : current altitude above Earth (see Formula 1 above)  
$\kappa$ : climb ratio parameter controlling the rate of transition from vertical to prograde flight
</div>

### Total/Net Acceleration

The total (or net) acceleration of the spacecraft is obtained by adding together all acceleration vectors currently acting on it. This includes contributions from gravity, atmospheric drag, and engine thrust.  
In our model, the gravitational acceleration always includes contributions from both the Earth and the Moon. Each body exerts its own gravitational acceleration on the spacecraft, and these are summed to give the total gravitational effect.

<div style="padding-left: 40px; max-width: 400px;">
    <div style="display: flex; justify-content: space-between; margin-top: 15px">
        <span>$\displaystyle \mathbf{a} = \mathbf{a}_{g,\text{Earth}} + \mathbf{a}_{g,\text{Moon}} + \mathbf{a}_d + \mathbf{a}_t$</span>
        <span style="color: #1a4cff; font-weight: bold;">(7)</span>
    </div>
</div>

### Time Integration

Once the net acceleration of the spacecraft has been computed, we must determine how its position and velocity change over time. This process is known as time integration.

At each time step, the spacecraft’s velocity is updated based on the current acceleration, and its position is then updated based on the updated velocity. This reflects the physical relationship that acceleration causes changes in velocity, and velocity causes changes in position.

In our model, we use a simple semi-implicit (symplectic) Euler method. This means that we first update the velocity using the current acceleration, and then update the position using the updated velocity. This approach is straightforward to implement and provides reasonable stability for the types of trajectories considered in this assignment.

By repeatedly applying this update process over small time steps, we simulate the continuous motion of the spacecraft through space and time.

<div style="padding-left: 40px; max-width: 400px;">
    <div style="display: flex; justify-content: space-between; margin-top: 15px">
        <span>$\displaystyle \mathbf{v}_\text{new} = \mathbf{v}_\text{old} + \mathbf{a} \,\Delta t$</span>
        <span style="color: #1a4cff; font-weight: bold;">(8)</span>
    </div>
    <div style="display: flex; justify-content: space-between; margin-top: 15px">
        <span>$\displaystyle \mathbf{r}_\text{new} = \mathbf{r}_\text{old} + \mathbf{v}_\text{new} \,\Delta t$</span>
        <span style="color: #1a4cff; font-weight: bold;">(9)</span>
    </div>
</div>

<div>
<div>where:</div>
<div style="padding-left:2em">
$\mathbf{r}_\text{new}$ : new spacecraft position vector  

$\mathbf{v}_\text{new}$ : new spacecraft velocity vector  
$\mathbf{r}_\text{old}$ : previous spacecraft position vector  
$\mathbf{v}_\text{old}$ : previous spacecraft velocity vector  
$\mathbf{a}$ : total acceleration acting on the spacecraft (see Formula 7 above)  
$\Delta t$ : simulation time step

### Ascent and Insertion into Orbit Phase

At <tt>+0d 00:00:00</tt> (i.e. at T minus $0$ seconds), we execute the first mission event, which is a <tt><b>LIFTOFF</b></tt>. 
As explained in the table above, we set <tt>spacecraft.engine_mode</tt> to <tt>'Gravity Turn'</tt> and set <tt>spacecraft.engine_acceleration</tt> to a value corresponding to the spacecraft’s initial engine configuration.
For the Artemis II mission, this means the acceleration resulting from combining the Orion Core Stage engines with its two Solid Rocket Boosters (SRBs). 
We also set <tt>spacecraft.drag_parameter</tt> consistent with this Core + SRB configuration, and initialise <tt>spacecraft.position</tt> and <tt>spacecraft.velocity</tt> based on the launch location and time.

The spacecraft will then remain in that engine and drag configuration until the next mission event occurs.
In our example above, this is a <tt><b>SEPARATION</b></tt> event that happens at <tt>+0d 00:02:08</tt> mission elapsed time, when we jettison the now-depleted Solid Rocket Boosters and continue with just the Orion Core Stage engines.
At that point, we now have a new <tt>engine_acceleration_magnitude</tt> ($0.01815$ in our example), along with a new <tt>drag_parameter</tt>, since the cross-sectional area has decreased but the mass is now significantly lower.

We then continue the simulation in that configuration until <tt>+0d 00:08:10</tt> into the mission, at which point we execute a <tt><b>BURN_CUTOFF</b></tt> event. 
This crucial cutoff event is referred to by NASA as ***Main Engine Cutoff (MECO)*** (see Figure 1b). 
This means we set <tt>spacecraft.engine_mode</tt> to <tt>'Off'</tt>, so the engines stop producing thrust and the spacecraft begins to coast, moving under its current velocity and gravitational acceleration alone.

Very soon after that, at <tt>+0d 00:08:18</tt>, we jettison the Core Stage, leaving only the Upper (ICPS) Stage. This again changes the drag parameter and also the engine acceleration parameter (even though the engine is not currently in use).
By this stage, we are well above the Earth’s atmosphere, so the air density is effectively zero, and the drag acceleration therefore also becomes zero.


### Earth Orbit Phase

Unlike on Earth, where drag and friction are significant, in the vacuum of space motion can continue indefinitely without energy loss due to drag (i.e. perpetual motion!). 
However, due to gravity, this motion is typically not in a straight line, but rather follows an elliptical orbit around the celestial body providing the primary gravitational influence.
As an object orbits a celestial body in this way, there are two key points in the orbit.

The ***perigee*** is the point where the orbiting object is closest to the celestial body, and the ***apogee*** is the point where it is furthest away (see Figure 1a).
The speed of the orbiting body is not constant - it is fastest at perigee and slowest at apogee.

Our Orion spacecraft’s initial gravity turn burn is designed to *"insert"* it into a low Earth elliptical orbit with an apogee altitude of approximately $2239$ km (see Figure 2c).
Main Engine Cutoff (MECO) occurred at <tt>+0d 00:08:10</tt>, after which the spacecraft coasts to reach that apogee at approximately <tt>+0d 00:49:00</tt> mission elapsed time.
If we were to remain in that orbit without further engine burns, the resulting next perigee would be approximately $25$ km above the Earth’s surface. 
At that altitude, the spacecraft would re-enter the atmosphere, experience drag, lose orbital energy, and eventually crash back to Earth.

To prevent this, as we approach the first apogee at <tt>+0d 00:49:00</tt> (about $2239$ km altitude), we perform a $17$-second burn using the Upper (ICPS) Stage engines. 
This increases the spacecraft’s speed and energy, changing the shape of the orbit. 
The current apogee of the orbit remains approximately the same, but the next perigee is raised from $25$ km to around $183$ km altitude - safely above the atmosphere. 
This burn is called a Perigee Raise Burn (see Figure 1d).

It then takes approximately $55$ minutes to *coast* to that next perigee at ~$183$ km altitude. 
At <tt>+0d 01:44:48</tt>, near perigee, we perform an Apogee Raise Burn lasting $17$ minutes and $52$ seconds (see Figure 2a). 
This modifies the orbit again so that the next apogee exceeds $70{,}200$ km altitude, placing the spacecraft into a very *high Earth elliptical orbit*.

Soon after this burn, at <tt>+0d 03:24:18</tt>, we jettison the ICPS (Interim Cryogenic Propulsion Stage), leaving only the Orion Service Module.
It takes approximately $10$ hours to coast to this high apogee above $70{,}200$ km altitude (see Figure 2a). 
No burn occurs at that apogee, and after a further ~$12$ hours, the spacecraft returns to perigee near Earth (see Figure 2b). 
During this 24-hour orbit, the crew conducts system checks before committing to the final stage of the mission.

### TransLunar Insertion (TLI)

Near the next perigee at <tt>+1d 01:25:27</tt> (now in day 2), the crew performs the ***TransLunar Insertion (TLI) burn*** for $5$ minutes and $55$ seconds (see Figure 2c), placing the spacecraft on a trajectory toward the Moon. 
The spacecraft then *coasts* (unpowered) for approximately $3$ days as it heads toward the Moon (or, to be more precise, toward where the Moon will be).
Because the Moon is also moving in its orbit around the Earth, we must aim not at the Moon's current position, but at where it will be upon arrival several days later. 
If this timing is incorrect, the spacecraft will catastrophically miss the Moon entirely and continue away from Earth forever into outer space.

### Lunar Sphere of Influence (SOI)

After almost $4$ days, the spacecraft enters the Moon’s ***Sphere of Influence*** (SOI), where the Moon becomes the dominant gravitational influence.
If the spacecraft enters the Moon’s Sphere of Influence at the correct speed and angle, it will pass around the far side of the Moon and be gravitationally redirected back toward Earth (following a figure-eight trajectory). 
This forms a ***free return trajectory***, where the Moon’s gravity effectively *"slingshots"* the spacecraft back toward Earth without requiring any further engine burns (apart from very small trajectory correction burns if necessary).

### Reentry Phase

As the spacecraft returns toward Earth, the Earth once again becomes the dominant gravitational influence. 
However, the entry trajectory needs to be near perfect.
If the approach is too shallow, the spacecraft will bounce off the atmosphere and remain in orbit. 
If the approach is too steep, the deceleration will be too rapid, converting kinetic energy to heat faster than the spacecraft’s heat shields can withstand.
If the entry angle and speed are correct, the spacecraft will experience controlled deceleration as it skims through the upper atmosphere.
NASA defines an important milestone called **Entry Interface**, which marks the formal start of re-entry. 
This is triggered not by time, but when the spacecraft’s altitude drops below $100$ km.
At approximately $8$ km altitude, the drogue parachutes deploy, increasing drag and slowing the spacecraft further.
Finally, at around $3$ km altitude, the main parachutes deploy, further increasing drag and reducing the descent speed to less than $10$ metres per second to achieve a safe splashdown!

### Mission Elapsed Time

In the above explanations, we have used human-friendly mission elapsed time strings such as "+1d 01:13:00".
Internally, it is simpler to store actual mission elapsed time and target mission elapsed time for events as ***numbers*** rather than ***strings***.
Storing times as numbers allows us to add, subtract, and compare whether one time is less than or equal to another.
So, we provide functions to convert between these two formats:

- <tt>convert_to_seconds('+0d 01:01:05') returns 3665</tt>
- <tt>convert_seconds_to_string(3665) returns '+0d 01:01:05'</tt>

When a <tt>MissionEvent</tt> is created, for example:

- <tt>MissionEvent(mission_elapsed_time=convert_to_seconds('+0d 00:02:08'), event_kind=EventKind.SEPARATION, event_name='SRB (Solid Rocket Booster) Separation', drag_parameter=0.000038, engine_acceleration_magnitude=0.0175)</tt>

the <tt>mission_elapsed_time</tt> property of the mission event will be a number, not a string.

Note that we now have two different ways of measuring time: mission elapsed time and ephemeris time. Both are (floating-point) numbers measured in seconds, but one is relative to the start of the mission and the other is relative to the year 2000.
We need to use ephemeris time when using the SPICE toolkit, but we need mission elapsed time when dealing with mission events.
Fortunately, we can easily convert one to the other, provided we know the ephemeris time when the mission commenced:

- <tt>ephemeris_time = launch_ephemeris_time + mission_elapsed_time</tt>, or
- <tt>mission_elapsed_time = ephemeris_time - launch_ephemeris_time</tt>
