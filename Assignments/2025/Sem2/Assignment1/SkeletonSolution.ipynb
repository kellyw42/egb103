Newton's First Law of Motion states:

<blockquote><i>"An object at rest remains at rest, and an object in motion remains in motion at constant speed and in a straight line unless acted upon by an external force."</i></blockquote>

In **Task 3**, we assume that the ball does not collide with any other objects—such as other balls, cushions, or pockets—during the time period under consideration.
Therefore, if a ball is in motion, it will continue in a straight line. However, to reflect real-world behavior, we must account for the **rolling resistance force** caused by friction between the ball and the table cloth. This resistance does not alter the ball’s direction but gradually reduces its speed. We assume this rolling resistance results in a **constant rate of deceleration**, represented by the constant <code>Ball.deceleration</code>, which may vary depending on the specific pool table and balls used.

- **Position** is measured in **metres**.
- **Speed** is the rate of change of position, measured in **metres per second**.
- **Acceleration (or deceleration)** is the rate of change of speed, measured in **metres per second per second**.

For example, if at time \( t = 0 \), the ball's speed is **8 m/s**, and the deceleration rate is **0.9 m/s²**, then after **1 second**, the speed will have decreased to **7.1 m/s**.


Our goal is to determine the ball’s **position and velocity** after a given time period, assuming no collisions occur during that time.

The **speed** of the ball is the **magnitude** of its velocity vector.

To determine the ball’s future position, we consider two scenarios:

1. **The ball comes to a complete stop during the time period**, or  
2. **The ball is still in motion at the end of the time period**

![Image of Scenario 1](images/scenario1.png) ![Image of Scenario 2](images/scenario2.png)

Our first step then is to compute the time until the ball will stop, based on its speed at the beginning of the time period and its rate of deceleration.
We can then compare that time to the duration of the time period to determine if we are dealing with Scenario 1 or Scenario 2.
Next we want to determine (under both scenarios), what the speed will be at the end of the time period and how far it will travel during the time period.
The distance travelled during the time period (in both scenarios) will be the blue coloured area under the speed curve. 

For the time period when we are actually moving, the distance travelled will be the average speed (initial speed + final speed divided by 2) multiplied by the time moving.

Once we know how far the ball will travel, we next need to determine its new position (and our new velocity).

This change in position will be in the direction we are moving. To determine this direction we normalize the initial velocity vector (since our direction of travel will stay the same).
This normalized vector will be in the same direction as the initial velocity vector, but will have a normalized magnitude of $1$. 
To normalize a vector we divide the $x$ and $y$ components of that vector by the magnitude of that vector. 
If the vector we are trying to normalize has a magnitude of zero, i.e. is the zero vector $(0,0)$, then that normalization process would result in a divide by zero, so we have a special case when normalizing the zero vector where the result should be the zero vector.

Once we have this normalized direction vector we can use it to compute our new position vector and our new  velocity vector.
The new position vector will be the old position vector plus the distance travelled in the direction of the normalized direction vector.
The new velocity vector will be our new speed in the direction of the normalized direction vector.

Let's consider two examples. 

In both cases we will assume that the time period that we are analysing is $0.01$ seconds and the ball deceleration rate is $0.1$ $m/s^2$.

### Example 1

***initial position*** = (1.9, 0.6)

***initial velocity*** = (4.9, -0.1)

***initial speed*** = magnitude of **initial velocity** 

$= \sqrt{4.9^2 + (-0.1)^2} = 4.901020301937139$. 

***ball will stop*** after ***initial speed*** / ***deceleration***

$= 4.901020301937139/0.1 = 49.01020301937139$ seconds. 

This is greater than our time period of $0.01$ seconds, so the ball will be in motion for that entire time period (Scenario 2). 

***new speed*** = ***initial speed*** - ***deceleration*** $\times$ ***time period***

$= 4.901020301937139 - 0.1 \times 0.01 = 4.900020301937139$ m/s. 

***distance travelled*** = average speed $\times$ time moving 

$= (4.901020301937139 + 4.900020301937139)/2 \times 0.01 = 0.04900520301937139$ metres.

***normalized direction vector*** = normalized ***initial velocity***

$= \left({4.9 \over \sqrt{4.9^2 + (-0.1)^2}}, {-0.1 \over \sqrt{4.9^2 + (-0.1)^2}}\right) = (0.9997918184634461, -0.02040391466251931)$. 

***new velocity*** = ***new speed*** $\times$ ***normalized direction vector***

$= (4.900020301937139 \times 0.9997918184634461, 4.900020301937139 \times -0.02040391466251931) = (4.899000208181536, -0.09997959608533748)$

***new position*** = ***initial position*** + ***distance travelled*** $\times$ ***normalized direction vector***

$= (1.9, 0.6) + (0.04900520301937139 \times 0.9997918184634461, 0.04900520301937139 \times -0.02040391466251931) = (1.9489950010409076, 0.5990001020195733)$

## Resolving Ball-to-Ball Collisions
When two billiard balls collide, the interaction occurs along a specific direction called the line of impact—the straight line connecting the centres of the two balls at the moment of contact. All collision forces act along this line, and so do the resulting changes in velocity.

Although billiard balls are rigid, they undergo slight elastic deformation during a collision. As they compress and then restore their shape, elastic potential energy is briefly stored and then released, pushing the balls apart. This process happens in just a few milliseconds, during which the balls exert equal and opposite forces on each other.

### Impulse and Change in Velocity

In real-world collisions, the forces involved are extremely large and act over very short time intervals. Rather than trying to compute these forces directly, we use the concept of impulse, which captures the total effect of a force over time.

The impulse-momentum theorem tells us that the impulse applied to an object is equal to its change in momentum:

$$\vec{J} = m \: \Delta \vec{v}$$

This allows us to bypass the need to know the exact force or duration of the collision. Instead, we model the collision as instantaneous, and since the impulse acts only along the line of impact, this allows us to focus solely on the change in velocity along that line.

### Relative Speed Along the Line of Impact
To resolve the collision, we focus on the relative speed of approach between the two balls along the line of impact. This is the component of their relative velocity projected onto the normalized direction vector between their centres.

Let: $\vec{v}_1$, $\vec{v}_2$: initial velocity vectors of the two balls

$\hat{n}$: unit vector from the centre of ball 1 to ball 2 (the line of impact)

Then the relative speed (a scalar) is:
$$s = (\vec{v}_1 - \vec{v}_2) \cdot \hat{n}$$
 
This scalar $s$ tells us how fast the balls are moving toward (or away from) each other along the line of impact. If the balls are moving towards each other then s will be negative; or positive if they are moving away from each other. Collision will therefore only occur if the relative speed is negative.

### Coefficient of Restitution
The coefficient of restitution $e$ quantifies how elastic the collision is. It is defined as the ratio of the relative speed of separation after the collision to the relative speed of approach before the collision—both measured along the line of impact:

$$e = \frac{(\vec{v}_2' - \vec{v}_1') \cdot \hat{n}}{(\vec{v}_1 - \vec{v}_2) \cdot \hat{n}}$$

In a perfectly elastic collision $e = 1$, so if for example the relative speed of approach was $-2$ m/s then the relative speed of separation immediately after the collision would be $+2$ m/s. 
For billiard balls, $e$ typically ranges from $0.92$ to $0.98$, indicating nearly elastic collisions, but where some of the system's kinetic energy has been lost during the collision to other forms of energy such as sound and heat. 

### Computing the Scalar Impulse
Using the impulse-momentum relationship and the definition of restitution, we can derive the scalar impulse $j$ applied along the line of impact. Assuming both balls have equal mass $m$, the result simplifies to:

$$j = \frac{-(1 + e)}{2} s$$
where:
$$s = ( \vec{v}_1 − \vec{v}_2) \cdot \hat{n}$$
  is the relative speed (before collision) along the line of impact.
  
This impulse is then used to calculate the new velocities of the two balls (after the collision):

$$\vec{v}_1' = \vec{v}_1 + j \: \hat{n} \qquad \text{(new velocity of ball 1)}$$
$$\vec{v}_2' = \vec{v}_2 - j \, \hat{n} \qquad \text{(new velocity of ball 2)}$$

<img src="images/lineOfImpact.png" width="400px"/>
 
The impulses are equal in magnitude and opposite in direction, ensuring conservation of momentum and consistent with Newton’s Third Law.

<h1>Task 5: Predict Future Collisions</h1>
</div> 
<p></p>
The Task 4 code for resolving collisions can only be applied at the precise moment that a collision takes place. 
We therefore need a mechanism for determining which collision(s) will happen next and precisely when.
To do this we consider every ball possibly colliding with every other ball (currently on the table). 
We also consider each ball possibly colliding with each of the four cushions (top, bottom, left and right) and each of the six pockets.
In determining whether ball A will collide with ball B we assume each ball will continue on its current tragetory and will slow according to the know deceleration rate due to rolling resistance, 
i.e. we assume that neither ball will collide with any other ball, cushion or pocket between now and then.
If we compute the precise future time for all possible collisions, then we can determine which of those collision(s) will happen first. 
We say <i>"collision(s)"</i> (i.e. possibly plural) as some future collisions might happen at the exact same time.

## Might they collide?
To determine if/when ball A will collide with ball B, we first need to be able to predict the precise future positions of both balls at any given point in time in the future. 
To do that, we simply use the ball motion functions from Task 3, but don't actually move the balls, just predict what their future position would be at that point in time.
If we know the precise future position of both balls at any given point in time, we can then use those future positions to compute the gap between them at that time point.
To determine if ball A will collide with ball B in the future, we first check their current relative speed. 
If their relative speed is negative then the balls are getting closer to one another, so they *may* collide in the future.
If their relative speed is currently positive, then they are moving apart and will move even further apart in the future, so no possible collision.

# When will they be closest to one another?
Assuming the relative speed is negative, 
the next question we seek to answer is *when* in the future will they be at their *closest* and what will be the gap between them at that point?
Again, this is still assuming that they don't first collide with some other object between now and then. 
In this case we are even excluding worrying about whether these two balls might collide with one another before getting to this point of minimum gap.
If the extrapolated future minimum gap is less than or equal to zero, 
then clearly they will collide in the future and the actual collision will happen when the gap becomes zero and before it becomes negative.

It would be nice to compute this time of minimum gap analytically by computing a formula based on the currrent positions and velocities of the two balls. 
Unfortunately, due to the deceleration of both balls, the equations that we would need to solve are order 4 polynomials, and so difficult to solve analytically.
We therefore resort to finding the time of minimum gap *numerically* using a loop that attempts to incrementally narrow in on the time of minimum gap.
To do this we need to start with a time interval where we know that the point of minimum gap will be within that time interval. 
Unless the relative speed is negative, we don't even try to find this point of minimum gap because we know it isn't in the future.
Otherwise, assuming the relative speed is negative, 
then the point of minimum gap will be in the future, so we can use $t=0$ as our initial lower bound on the time interval during which we will reach minimum gap.
We can also for each ball determine (from Task 3) when it will eventually stop rolling. 
After both balls have stopped rolling, the balls can't possibly get any closer, so that provides our upper bound for the time interval during which we will reach the minimum gap.

To *numerically* find the time of minimum gap within this interval, we use a method known as the <a href="https://en.wikipedia.org/wiki/Golden-section_search"><b><i>Golden-section search</i></b></a>. 
This is a well-established technique for locating the minimum (or maximum) of a unimodal function—a function that has only one extremum within a given interval. 
It works by evaluating the function (in our case, the gap between the two balls) at two carefully chosen interior time points that divide the interval according to the golden ratio (approximately 1.618). 
This ratio ensures that with each iteration, we can discard the portion of the interval that cannot contain the minimum, while preserving the golden ratio structure in the remaining interval.
At each step, we compare the gap values at the two interior time points. 
If the gap is smaller at the earlier time point, we know the minimum must lie in the earlier portion of the time interval, so we update the upper bound. 
If the gap is smaller at the later time point, we update the lower bound instead. 

<img src="images/findmin.png" width="400px"/>

In the above example, the gap at interior time point 1 is less than the gap at time point 2, so time point 2 becomes our new upper bound.

This process is repeated, shrinking the interval each time, until it becomes so small that we can confidently say we’ve found the time of minimum gap to within a very fine tolerance.
After only a few dozen iterations, we will know the time of maximum gap to within less than one nanosecond (which is close enough for our modelling).

## When precisely will they collide?

Once we know the time of minimum gap we can determine the gap at that time and determine if there will be a collision based on whether than minimum gap is less or equal to zero.
If there will be a collision, we then need to determine precisely when that collision will occur which will happen when the gap between the balls first becomes zero.
We again use an iterative search approach, again starting with an upper and lower time bound on when the actual collision will occur.
Again, $t=0$ i.e. *now* provides the lower bound (provided the relative speed is negative), 
and the time of minimum gap (that we just computed) will provide an upper bound on when the collision will actually occur.
The collision might actually be happening right now (i.e. at time $t=0$), so before we start searching, we first check if the balls are already "touching", 
i.e. if the gap between them is already less than or equal to zero. 
Otherwise we follow an iterative approach, but this time somewhat simpler. Rather than dividing the current time interval into 3 parts, we just divide it into 2 equal time periods. 
We check the time point midway between the current upper bound and lower bound. 
If the gap between the balls is less than zero at that mid point then the collision must occur in the first half of the time interval, so we update our current upper bound.
Otherwise the collision must occur in the second half of the time interval so we update our current lower bound.

We examine the gap between the balls at the mid-point between the upper and lower time bounds and find that it is positive, which means it must become zero somewhere in the second half of the time interval, so the mid-point becomes our new lower bound.

As before, after only a few dozen iterations we know the precise time of collision to within one nanosecond (again, close enough for our modelling).

We now wish to create a visual animation of the movement and collisions of the balls after the que ball is struck by the que stick.
To create the animation, we need to take a series of snapshots of the current locations of the balls that will become the frames of our video.
Based on the frames per second that we desire, we can determine the time interval for each frame.
For example, if we want 100 frames per second, then our time interval will be 0.01 seconds.

For each frame we only need a snapshot of each ball's position, not it's velocity, so the create_frame function from the pool table module takes a list of balls and creates a list of tuples capturing the current location of each ball.

The animation begins by create a snapshot of the initial ball locations.
We then call function find_next_collisions to determine when the next collision will take place.
We then ask the question, which will come first, the next collision or the next frame.
For example, if the next collision will happen in 0.0324 seconds and the next frame will be in 0.01 seconds, then the next frame will come first.
If the next frame will come first, then we know there won't be any collisions in the time period leading up to the next frame so we can
fast forward time by moving all balls for a period of 0.01 seconds by using the move functions from Task 3.
If however the next collision will occur in say 0.00338 seconds, then that will occur before the next frame.
So, we instead fast forward time by only 0.00338 seconds until we come to the precise time point when that collision occurs.
We then resolve those next collision(s) using the functions from Task 4. 
Note, these collision resolution functions only change the velocities of the balls, not their positions.
Once those collision(s) are resolved, we then need to call find_next_collisions again because both the positions and velocities have changed since we last called it.
We again then ask, which comes first, the next frame or the next collision.
Note, in this case the next frame doesn't occur in 0.01 seconds but rather in 0.01 - 0.00338 = 0.00662 since we have already fast forwarded the first 0.00338 seconds of the current frame.
This continues until all balls are stationary, i.e. have a velocity of zero.

````python
# This code cell should define the following functions:
#     move_all_balls, resolve_collisions and simulate_motion_until_all_stopped

#
# Do not cut and paste any of those function definitions to duplicate or move them elsewhere
#
# Do not implement any other functions in this code cell.
#
# Do not include any global variable assignments or any test code in this code cell (include it in a seperate code cell below).
# i.e. this code cell should contain only function definitions, nothing else!
#
# -------------------------------------------------------------------------------------------------------------


# Function Overview: Moves all balls on the table forward in time by a given time interval, updating their positions and velocities.
#
# Makes use of: move
#
# Parameters:
#    balls (list): A list of Ball objects to be moved.
#    time_delta (float): The time interval over which to move the balls.
#
# Returns:
#    None: Updates each ball's position and velocity attributes in place.
#
def move_all_balls(balls, time_delta):
    # This function has been implemented for you, do not modify it in any way
    for ball in balls:
        move(ball, time_delta)


# Function Overview: Resolves all collisions that are scheduled to occur at the current time step.
#
# Makes use of: resolve_collision
#
# Parameters:
#    next_collisions (list): A list of (object, ball) pairs involved in collisions.
#    balls (list): A list of all Ball objects on the table.
#
# Returns:
#    None: Updates the velocity of the balls involved in collisions.
#
def resolve_collisions(next_collisions, balls):
    # This function has been implemented for you, do not modify it in any way
    for (object, ball) in next_collisions:
        object.resolve_collision(ball, balls)


# Function Overview: Simulates the motion of all balls on the table until they come to a complete stop,
#                    recording frames at regular intervals for animation or visualization.
#
# Makes use of: find_next_collisions, move_all_balls, resolve_collisions, all_stationary, create_frame
#
# Parameters:
#    balls (list): A list of Ball objects to simulate.
#    time_delta (float): The time interval between animation frames.
#
# Returns:
#    list: A list of frames representing the state of the table at each recorded time step.
#
def simulate_motion_until_all_stopped(balls, time_delta):
    # This function has been implemented for you, do not modify it in any way
    frames = []
    time_until_next_collision, next_collisions = find_next_collisions(balls)
    time_until_next_frame = 0
    while frames == [] or not all_stationary(balls):
        if time_until_next_frame == 0:
            frames.append(pool_table.create_frame(balls))
            time_until_next_frame = time_delta
        if next_collisions and time_until_next_collision < time_until_next_frame:
            move_all_balls(balls, time_until_next_collision)
            time_until_next_frame -= time_until_next_collision
            resolve_collisions(next_collisions, balls)
            time_until_next_collision, next_collisions = find_next_collisions(balls)
        else:
            move_all_balls(balls, time_until_next_frame)
            time_until_next_collision -= time_until_next_frame
            time_until_next_frame = 0
    return frames
````
